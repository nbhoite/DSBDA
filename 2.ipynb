***
**Step 1:** Importing the Python Libraries
***

```python
import pandas as pd
import numpy as np
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

***
**Step 2:** Reading CSV file into Data Frames
***

```python
df=pd.read_csv("AcademicPerformance.csv")
```

```python
df
```

***
**Step 3:** Data Preprocessing
***

```python
series = pd.isnull(df["Math_Score"])
df[series]
```

```python
df['Math_Score'] = df['Math_Score'].fillna(df['Math_Score'].mean())
df
```

```python
series = pd.isnull(df["Science_Score"])
df[series]
```

```python
df['Science_Score'] = df['Science_Score'].fillna(df['Science_Score'].mean())
df
```

***
**Step 4:** Finding Outliers
***

```python
z = np.abs(stats.zscore(df['Math_Score']))
z
```

```python
threshold = 0.17
```

```python
outliers = np.where(z<threshold)
for i in outliers:
    df.drop(i,inplace=True)
```
