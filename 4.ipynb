**STEP 4**: Linear Regression Modeling
***

```python
bins=(0,4,25)
group_names=["bad_railway_access", "good_railway_access"]
df["Railway_Index"] = pd.cut(df.RAD, bins = bins, labels = group_names)
df["Railway_Index"]
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
label_quality=LabelEncoder()
df["Railway_Index"]=label_quality.fit_transform(df["Railway_Index"])
df.head(10)
```

```python
sns.countplot(x=df["Railway_Index"])
```

```python
df.astype('int32').dtypes
df.head()
```

```python
X=df.drop("Railway_Index",axis=1)
y=df['Railway_Index']
```

```python
from sklearn.ensemble import  RandomForestClassifier
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train , y_test= train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
X_train[:1]
```

```python
rfc=RandomForestClassifier(n_estimators=100)
rfc.fit(X_train, y_train)
pred_rfc=rfc.predict(X_test)
pred_rfc[:2]
```

```python
from sklearn.metrics import accuracy_score
cm=accuracy_score(y_test,pred_rfc)
cm
```
